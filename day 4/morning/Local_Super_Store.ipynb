# Global Superstore Dataset Report

```python
# ============================================================
# GENERATE PDF REPORT - OUTPUT ONLY
# ============================================================

!pip install -q reportlab

import os
import math
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from reportlab.lib import colors
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table,
    TableStyle,
    PageBreak,
    Image
)
from google.colab import files

# ------------------------------------------------------------
# FILE NAME
# ------------------------------------------------------------

pdf_file = "Global_Superstore_Data_Report.pdf"

# ------------------------------------------------------------
# PDF SETTINGS
# ------------------------------------------------------------

doc = SimpleDocTemplate(
    pdf_file,
    pagesize=A4,
    rightMargin=40,
    leftMargin=40,
    topMargin=45,
    bottomMargin=45
)

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "ReportTitle",
    parent=styles["Title"],
    alignment=TA_CENTER,
    fontSize=22,
    leading=28,
    spaceAfter=20
)

subtitle_style = ParagraphStyle(
    "Subtitle",
    parent=styles["Normal"],
    alignment=TA_CENTER,
    fontSize=11,
    leading=16,
    spaceAfter=30
)

heading_style = ParagraphStyle(
    "Heading",
    parent=styles["Heading1"],
    fontSize=15,
    leading=20,
    spaceBefore=12,
    spaceAfter=10
)

subheading_style = ParagraphStyle(
    "SubHeading",
    parent=styles["Heading2"],
    fontSize=12,
    leading=16,
    spaceBefore=10,
    spaceAfter=8
)

normal_style = ParagraphStyle(
    "NormalReport",
    parent=styles["Normal"],
    fontSize=9.5,
    leading=14,
    spaceAfter=7
)

small_style = ParagraphStyle(
    "Small",
    parent=styles["Normal"],
    fontSize=7.5,
    leading=10
)

# ------------------------------------------------------------
# HELPER FUNCTIONS
# ------------------------------------------------------------

story = []


def add_heading(text):
    story.append(Paragraph(text, heading_style))
    story.append(Spacer(1, 5))


def add_subheading(text):
    story.append(Paragraph(text, subheading_style))


def add_text(text):
    story.append(Paragraph(text, normal_style))


def dataframe_to_table(dataframe, max_rows=None):

    data = dataframe.copy()

    if max_rows is not None:
        data = data.head(max_rows)

    # Convert values to readable strings
    formatted = []

    for value in data.columns:
        formatted.append(str(value))

    table_data = [formatted]

    for _, row in data.iterrows():

        row_data = []

        for value in row:

            if pd.isna(value):
                value = "NULL"

            elif isinstance(value, (float, np.floating)):
                value = f"{value:,.2f}"

            elif isinstance(value, (int, np.integer)):
                value = f"{value:,}"

            else:
                value = str(value)

            row_data.append(
                Paragraph(value, small_style)
            )

        table_data.append(row_data)

    # Calculate column widths
    page_width = A4[0] - 80

    column_count = len(data.columns)

    if column_count > 0:
        col_width = page_width / column_count
        col_widths = [col_width] * column_count
    else:
        col_widths = None

    table = Table(
        table_data,
        colWidths=col_widths,
        repeatRows=1
    )

    table.setStyle(
        TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.lightgrey),
            ("TEXTCOLOR", (0, 0), (-1, 0), colors.black),
            ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
            ("FONTSIZE", (0, 0), (-1, 0), 7.5),
            ("ALIGN", (0, 0), (-1, 0), "CENTER"),
            ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
            ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
            ("ROWBACKGROUNDS", (0, 1), (-1, -1),
             [colors.white, colors.whitesmoke]),
            ("LEFTPADDING", (0, 0), (-1, -1), 4),
            ("RIGHTPADDING", (0, 0), (-1, -1), 4),
            ("TOPPADDING", (0, 0), (-1, -1), 4),
            ("BOTTOMPADDING", (0, 0), (-1, -1), 4),
        ])
    )

    return table


def add_chart(filename, width=6.5*inch):
    if os.path.exists(filename):
        img = Image(filename)
        img.drawWidth = width
        img.drawHeight = width * 0.55
        story.append(img)
        story.append(Spacer(1, 10))


def save_chart(filename):
    plt.tight_layout()
    plt.savefig(
        filename,
        dpi=150,
        bbox_inches="tight"
    )
    plt.close()


# ------------------------------------------------------------
# CALCULATIONS
# ------------------------------------------------------------

rows, columns = df.shape

duplicate_rows = df.duplicated().sum()
duplicate_row_ids = df["Row ID"].duplicated().sum()

invalid_dates = len(
    df[df["Ship Date"] < df["Order Date"]]
)

missing_postal = df["Postal Code"].isna().sum()

missing_shipping_days = (
    df["Shipping Days"].isna().sum()
    if "Shipping Days" in df.columns
    else "N/A"
)

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

average_sales = df["Sales"].mean()
average_profit = df["Profit"].mean()

# ------------------------------------------------------------
# TITLE
# ------------------------------------------------------------

story.append(Spacer(1, 50))

story.append(
    Paragraph(
        "Global Superstore Dataset Report",
        title_style
    )
)

story.append(
    Paragraph(
        "Data Quality Analysis, Transformation Validation and Exploratory Analysis",
        subtitle_style
    )
)

add_text(
    "<b>Dataset Source:</b> Kaggle - Global Superstore"
)

add_text(
    "<b>Report Purpose:</b> To analyse the structure, quality, "
    "transformations and important characteristics of the cleaned dataset."
)

story.append(PageBreak())

# ------------------------------------------------------------
# 1. DATASET OVERVIEW
# ------------------------------------------------------------

add_heading("1. Dataset Overview")

overview = pd.DataFrame({
    "Metric": [
        "Number of Records",
        "Number of Columns",
        "Duplicate Rows",
        "Duplicate Row IDs",
        "Countries",
        "Markets",
        "Regions",
        "Product Categories",
        "Earliest Order Date",
        "Latest Order Date"
    ],
    "Value": [
        rows,
        columns,
        duplicate_rows,
        duplicate_row_ids,
        df["Country"].nunique(),
        df["Market"].nunique(),
        df["Region"].nunique(),
        df["Category"].nunique(),
        str(df["Order Date"].min().date()),
        str(df["Order Date"].max().date())
    ]
})

story.append(dataframe_to_table(overview))
story.append(Spacer(1, 15))

add_text(
    f"The dataset contains <b>{rows:,} records</b> and "
    f"<b>{columns} columns</b>. It contains information about "
    "customers, products, locations, sales, profits, shipping and order priority."
)

# ------------------------------------------------------------
# 2. COLUMN INFORMATION
# ------------------------------------------------------------

add_heading("2. Column Information")

column_info = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Non-Null": df.notna().sum().values,
    "Missing": df.isna().sum().values,
    "Unique": df.nunique().values
})

story.append(dataframe_to_table(column_info))
story.append(PageBreak())

# ------------------------------------------------------------
# 3. MISSING VALUE ANALYSIS
# ------------------------------------------------------------

add_heading("3. Missing Value Analysis")

missing = df.isna().sum()
missing_percentage = (missing / len(df)) * 100

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing.values,
    "Missing %": missing_percentage.round(2).values
})

missing_report = missing_report.sort_values(
    by="Missing Values",
    ascending=False
)

story.append(dataframe_to_table(missing_report))

add_text(
    f"The main missing-value issue is the <b>Postal Code</b> column, "
    f"which contains <b>{missing_postal:,}</b> missing values. "
    "These values were retained because reliable postal codes cannot "
    "be safely inferred from the available data."
)

# Missing chart
missing_nonzero = missing_report[
    missing_report["Missing Values"] > 0
]

if len(missing_nonzero) > 0:

    plt.figure(figsize=(9, 4))

    plt.bar(
        missing_nonzero["Column"],
        missing_nonzero["Missing Values"]
    )

    plt.title("Missing Values by Column")
    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=60, ha="right")

    save_chart("/content/missing_values.png")

    add_chart("/content/missing_values.png")

# ------------------------------------------------------------
# 4. DUPLICATE ANALYSIS
# ------------------------------------------------------------

add_heading("4. Duplicate Analysis")

duplicate_report = pd.DataFrame({
    "Check": [
        "Duplicate complete rows",
        "Duplicate Row IDs"
    ],
    "Count": [
        duplicate_rows,
        duplicate_row_ids
    ]
})

story.append(dataframe_to_table(duplicate_report))

add_text(
    "No duplicate records were identified. Row ID values were also "
    "checked and found to be unique."
)

# ------------------------------------------------------------
# 5. NUMERICAL DATA ANALYSIS
# ------------------------------------------------------------

add_heading("5. Numerical Data Analysis")

numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Cost"
]

if "Shipping Days" in df.columns:
    numeric_columns.append("Shipping Days")

numeric_summary = df[numeric_columns].describe().T

numeric_summary["Median"] = df[numeric_columns].median()
numeric_summary["Missing"] = df[numeric_columns].isna().sum()

numeric_summary = numeric_summary.reset_index()
numeric_summary = numeric_summary.rename(
    columns={"index": "Variable"}
)

story.append(dataframe_to_table(numeric_summary))

add_text(
    "The numerical attributes were inspected for unusual or invalid "
    "values. Negative Profit values were retained because they represent "
    "legitimate business losses rather than data-entry errors."
)

story.append(PageBreak())

# ------------------------------------------------------------
# 6. CATEGORICAL DATA ANALYSIS
# ------------------------------------------------------------

add_heading("6. Categorical Data Analysis")

categorical_columns = [
    "Ship Mode",
    "Segment",
    "Category",
    "Sub-Category",
    "Market",
    "Region",
    "Order Priority"
]

for column in categorical_columns:

    if column in df.columns:

        add_subheading(column)

        counts = df[column].value_counts()

        category_table = pd.DataFrame({
            "Value": counts.index.astype(str),
            "Count": counts.values,
            "Percentage": (
                counts.values / len(df) * 100
            ).round(2)
        })

        story.append(
            dataframe_to_table(category_table)
        )

        story.append(Spacer(1, 8))

# ------------------------------------------------------------
# 7. DATE ANALYSIS
# ------------------------------------------------------------

add_heading("7. Date Analysis")

date_report = pd.DataFrame({
    "Metric": [
        "Earliest Order Date",
        "Latest Order Date",
        "Number of Years",
        "Number of Months",
        "Invalid Shipping Dates"
    ],
    "Value": [
        df["Order Date"].min().date(),
        df["Order Date"].max().date(),
        df["Order Date"].dt.year.nunique(),
        df["Order Date"].dt.to_period("M").nunique(),
        invalid_dates
    ]
})

story.append(dataframe_to_table(date_report))

# Orders by year chart
orders_by_year = (
    df["Order Date"]
    .dt.year
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 4))

plt.bar(
    orders_by_year.index.astype(str),
    orders_by_year.values
)

plt.title("Number of Orders by Year")
plt.xlabel("Year")
plt.ylabel("Number of Orders")

save_chart("/content/orders_by_year.png")

add_chart("/content/orders_by_year.png")

# ------------------------------------------------------------
# 8. TRANSFORMATION SUMMARY
# ------------------------------------------------------------

add_heading("8. Data Transformation Summary")

transformation_summary = pd.DataFrame({
    "Transformation / Check": [
        "Order Date conversion",
        "Ship Date conversion",
        "Duplicate row check",
        "Duplicate Row ID check",
        "Invalid date check",
        "Missing Postal Code handling",
        "Numerical data validation",
        "Categorical data validation",
        "Negative Profit handling",
        "Shipping Days creation"
    ],
    "Result": [
        "Converted to datetime",
        "Converted to datetime",
        "No duplicates found",
        "No duplicates found",
        "No invalid dates found",
        "Missing values retained as NULL",
        "Values checked",
        "No obvious inconsistencies found",
        "Retained as valid business losses",
        "Ship Date - Order Date"
    ]
})

story.append(
    dataframe_to_table(transformation_summary)
)

add_text(
    "The transformation stage focused on improving data consistency "
    "and validating the dataset without unnecessarily removing valid records."
)

story.append(PageBreak())

# ------------------------------------------------------------
# 9. SHIPPING DAYS ANALYSIS
# ------------------------------------------------------------

add_heading("9. Shipping Days Analysis")

if "Shipping Days" in df.columns:

    shipping_summary = df["Shipping Days"].describe().to_frame().T

    story.append(
        dataframe_to_table(shipping_summary.reset_index())
    )

    plt.figure(figsize=(8, 4))

    plt.hist(
        df["Shipping Days"].dropna(),
        bins=range(
            int(df["Shipping Days"].min()),
            int(df["Shipping Days"].max()) + 2
        )
    )

    plt.title("Distribution of Shipping Days")
    plt.xlabel("Shipping Days")
    plt.ylabel("Number of Records")

    save_chart("/content/shipping_days.png")

    add_chart("/content/shipping_days.png")

# ------------------------------------------------------------
# 10. SALES BY CATEGORY
# ------------------------------------------------------------

add_heading("10. Sales by Category")

sales_by_category = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales_table = sales_by_category.reset_index()
category_sales_table.columns = [
    "Category",
    "Total Sales"
]

story.append(
    dataframe_to_table(category_sales_table)
)

plt.figure(figsize=(8, 4))

plt.bar(
    sales_by_category.index,
    sales_by_category.values
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=30)

save_chart("/content/sales_category.png")

add_chart("/content/sales_category.png")

# ------------------------------------------------------------
# 11. PROFIT BY CATEGORY
# ------------------------------------------------------------

add_heading("11. Profit by Category")

profit_by_category = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

profit_table = profit_by_category.reset_index()
profit_table.columns = [
    "Category",
    "Total Profit"
]

story.append(
    dataframe_to_table(profit_table)
)

plt.figure(figsize=(8, 4))

plt.bar(
    profit_by_category.index,
    profit_by_category.values
)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=30)

save_chart("/content/profit_category.png")

add_chart("/content/profit_category.png")

# ------------------------------------------------------------
# 12. SALES BY MARKET
# ------------------------------------------------------------

add_heading("12. Sales by Market")

sales_by_market = (
    df.groupby("Market")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

market_table = sales_by_market.reset_index()
market_table.columns = [
    "Market",
    "Total Sales"
]

story.append(
    dataframe_to_table(market_table)
)

plt.figure(figsize=(8, 4))

plt.bar(
    sales_by_market.index,
    sales_by_market.values
)

plt.title("Total Sales by Market")
plt.xlabel("Market")
plt.ylabel("Sales")
plt.xticks(rotation=30)

save_chart("/content/sales_market.png")

add_chart("/content/sales_market.png")

story.append(PageBreak())

# ------------------------------------------------------------
# 13. SALES BY REGION
# ------------------------------------------------------------

add_heading("13. Sales by Region")

sales_by_region = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

region_table = sales_by_region.reset_index()
region_table.columns = [
    "Region",
    "Total Sales"
]

story.append(
    dataframe_to_table(region_table)
)

plt.figure(figsize=(9, 5))

plt.bar(
    sales_by_region.index,
    sales_by_region.values
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=70)

save_chart("/content/sales_region.png")

add_chart("/content/sales_region.png")

# ------------------------------------------------------------
# 14. TOP 10 PRODUCTS
# ------------------------------------------------------------

add_heading("14. Top 10 Products by Sales")

top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_table = top_products.reset_index()
top_products_table.columns = [
    "Product Name",
    "Total Sales"
]

story.append(
    dataframe_to_table(top_products_table)
)

plt.figure(figsize=(9, 5))

plt.barh(
    top_products.index[::-1],
    top_products.values[::-1]
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

save_chart("/content/top_products.png")

add_chart("/content/top_products.png")

# ------------------------------------------------------------
# 15. SHIP MODE ANALYSIS
# ------------------------------------------------------------

add_heading("15. Ship Mode Analysis")

ship_mode_counts = df["Ship Mode"].value_counts()

ship_mode_table = ship_mode_counts.reset_index()
ship_mode_table.columns = [
    "Ship Mode",
    "Number of Orders"
]

story.append(
    dataframe_to_table(ship_mode_table)
)

plt.figure(figsize=(8, 4))

plt.bar(
    ship_mode_counts.index,
    ship_mode_counts.values
)

plt.title("Orders by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)

save_chart("/content/ship_mode.png")

add_chart("/content/ship_mode.png")

# ------------------------------------------------------------
# 16. ORDER PRIORITY
# ------------------------------------------------------------

add_heading("16. Order Priority Analysis")

priority_counts = df["Order Priority"].value_counts()

priority_table = priority_counts.reset_index()
priority_table.columns = [
    "Order Priority",
    "Number of Orders"
]

story.append(
    dataframe_to_table(priority_table)
)

# ------------------------------------------------------------
# 17. KEY STATISTICS
# ------------------------------------------------------------

add_heading("17. Key Statistics")

key_statistics = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity Sold",
        "Average Sales per Record",
        "Average Profit per Record",
        "Average Shipping Days"
    ],
    "Value": [
        f"{total_sales:,.2f}",
        f"{total_profit:,.2f}",
        f"{total_quantity:,}",
        f"{average_sales:,.2f}",
        f"{average_profit:,.2f}",
        f"{df['Shipping Days'].mean():,.2f}"
        if "Shipping Days" in df.columns else "N/A"
    ]
})

story.append(
    dataframe_to_table(key_statistics)
)

story.append(PageBreak())

# ------------------------------------------------------------
# 18. KEY FINDINGS
# ------------------------------------------------------------

add_heading("18. Key Findings")

best_sales_category = sales_by_category.idxmax()
best_profit_category = profit_by_category.idxmax()
best_market = sales_by_market.idxmax()
best_region = sales_by_region.idxmax()
best_ship_mode = ship_mode_counts.idxmax()

findings = [
    f"The dataset contains <b>{rows:,} records</b> and <b>{columns} columns</b>.",

    f"No duplicate rows or duplicate Row IDs were identified.",

    f"The Postal Code column contains <b>{missing_postal:,}</b> missing values. "
    "These were retained as NULL because they could not be reliably inferred.",

    f"The <b>{best_sales_category}</b> category generated the highest total sales.",

    f"The <b>{best_profit_category}</b> category generated the highest total profit.",

    f"The <b>{best_market}</b> market generated the highest total sales.",

    f"The <b>{best_region}</b> region generated the highest total sales.",

    f"The most frequently used shipping mode was <b>{best_ship_mode}</b>.",

    "Negative Profit values were retained because they represent legitimate business losses.",

    "Shipping Days was created by calculating the difference between Ship Date and Order Date."
]

for i, finding in enumerate(findings, 1):

    add_text(f"<b>{i}.</b> {finding}")

# ------------------------------------------------------------
# 19. FINAL DATA QUALITY SUMMARY
# ------------------------------------------------------------

add_heading("19. Final Data Quality Summary")

final_quality = pd.DataFrame({
    "Quality Check": [
        "Duplicate Rows",
        "Duplicate Row IDs",
        "Invalid Shipping Dates",
        "Missing Postal Codes",
        "Missing Shipping Days",
        "Records After Transformation"
    ],
    "Result": [
        duplicate_rows,
        duplicate_row_ids,
        invalid_dates,
        missing_postal,
        missing_shipping_days,
        len(df)
    ]
})

story.append(
    dataframe_to_table(final_quality)
)

# ------------------------------------------------------------
# 20. CONCLUSION
# ------------------------------------------------------------

add_heading("20. Conclusion")

add_text(
    "The Global Superstore dataset was successfully extracted and transformed "
    "for further analysis and data warehousing."
)

add_text(
    "The dataset was checked for duplicate records, missing values, invalid "
    "dates, numerical inconsistencies and categorical inconsistencies. "
    "The Order Date and Ship Date fields were converted to appropriate "
    "datetime formats."
)

add_text(
    "Missing Postal Code values were retained because reliable values could "
    "not be inferred. Negative Profit values were also retained because they "
    "represent legitimate business losses."
)

add_text(
    "A new Shipping Days attribute was created using the difference between "
    "Ship Date and Order Date."
)

add_text(
    "The analysis provided insights into sales, profit, categories, markets, "
    "regions, products, shipping methods and order priorities."
)

add_text(
    "<b>The cleaned dataset is now ready for the Load stage of the ETL pipeline.</b>"
)

# ------------------------------------------------------------
# BUILD PDF
# ------------------------------------------------------------

doc.build(story)

print("PDF report generated successfully!")
print(f"File: {pdf_file}")

# ------------------------------------------------------------
# AUTOMATIC DOWNLOAD
# ------------------------------------------------------------

files.download(pdf_file)
```
